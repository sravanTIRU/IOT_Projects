```python
import random
import time
from datetime import datetime
import pandas as pd
import matplotlib.pyplot as plt

print("=== IoT NETWORK LAYER ===")
print("Sensor Data Transmission System\n")

network_data = []

for i in range(10):
    temperature = round(random.uniform(25, 40), 2)
    humidity = round(random.uniform(40, 80), 2)
    light = random.randint(100, 1000)

    # Simulate network transmission
    latency = round(random.uniform(10, 100), 2)
    packet_size = random.randint(50, 150)

    if random.random() > 0.1:
        status = "Success"
    else:
        status = "Failed"

    data = {
        "Time": datetime.now().strftime("%H:%M:%S"),
        "Temperature": temperature,
        "Humidity": humidity,
        "Light": light,
        "Packet Size (Bytes)": packet_size,
        "Latency (ms)": latency,
        "Status": status
    }

    network_data.append(data)

    print(f"Packet {i + 1}")
    print(f"Temperature : {temperature} °C")
    print(f"Humidity    : {humidity} %")
    print(f"Light       : {light} lux")
    print(f"Packet Size : {packet_size} bytes")
    print(f"Latency     : {latency} ms")
    print(f"Status      : {status}")
    print("---------------------------")

    time.sleep(1)

print("Network Data Transmission Completed!")

# Display network data
df = pd.DataFrame(network_data)

print("\nIoT Network Layer Data")
display(df)

# Graph report
fig, axes = plt.subplots(2, 1, figsize=(10, 7))

axes[0].plot(df.index + 1, df["Latency (ms)"], marker="o")
axes[0].set_title("Network Latency Monitoring")
axes[0].set_ylabel("Latency (ms)")
axes[0].set_xlabel("Packet Number")
axes[0].grid(True)

axes[1].bar(
    df.index + 1,
    df["Packet Size (Bytes)"]
)
axes[1].set_title("Packet Size Monitoring")
axes[1].set_ylabel("Packet Size (Bytes)")
axes[1].set_xlabel("Packet Number")
axes[1].grid(True)

plt.tight_layout()
plt.show()

# Final output
print("\n=== NETWORK CONDITION MONITORING ===")

for reading in network_data:
    if reading["Status"] == "Failed":
        print("Packet Transmission Failed")
    elif reading["Latency (ms)"] > 70:
        print("High Network Latency")
    else:
        print("Network Condition: Normal")
```

```python
import random
import pandas as pd
import matplotlib.pyplot as plt

print("=== OSI DATA LINK LAYER ===")
print("Virtual Frame Monitoring System\\n")

frame_data = []
devices = ["AA:10:BC:20:01:01", "AA:10:BC:20:01:02",
           "AA:10:BC:20:01:03", "AA:10:BC:20:01:04"]

for i in range(10):
    source = random.choice(devices)
    destination = random.choice([d for d in devices if d != source])
    frame_size = random.randint(64, 1500)
    checksum_ok = random.random() > 0.15
    status = "Valid" if checksum_ok else "Error"

    data = {
        "Frame": i + 1,
        "Source MAC": source,
        "Destination MAC": destination,
        "Frame Size": frame_size,
        "Checksum": "OK" if checksum_ok else "Error",
        "Status": status
    }
    frame_data.append(data)

    print(f"Frame {i + 1}")
    print(f"Source MAC      : {source}")
    print(f"Destination MAC : {destination}")
    print(f"Frame Size      : {frame_size} bytes")
    print(f"Status          : {status}")
    print("---------------------------")

print("Data link layer monitoring completed!")
```

```python
df = pd.DataFrame(frame_data)

print("Data Link Layer Data")
display(df)
plt.figure(figsize=(10, 4))
plt.bar(df["Frame"], df["Frame Size"])
plt.title("Frame Size Monitoring")
plt.ylabel("Frame Size (bytes)")
plt.xlabel("Frame Number")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
print("=== FRAME CONDITION MONITORING ===")

for reading in frame_data:
    if reading["Status"] == "Error":
        print(f"Frame {reading['Frame']} -> ERROR DETECTED")
    else:
        print(f"Frame {reading['Frame']} -> NORMAL")

```

```python
import random
import pandas as pd
import matplotlib.pyplot as plt

print("=== OSI TRANSPORT LAYER ===")
print("Virtual Segment Monitoring System\\n")

transport_data = []

for i in range(10):
    protocol = random.choice(["TCP", "UDP"])
    port = random.choice([80, 443, 1883, 5000])
    segment_size = random.randint(100, 1400)
    retries = random.randint(0, 3) if protocol == "TCP" else 0
    delay = round(random.uniform(5, 100) + retries * 10, 2)
    status = "Delivered" if random.random() > 0.1 else "Lost"

    data = {
        "Segment": i + 1,
        "Protocol": protocol,
        "Port": port,
        "Segment Size": segment_size,
        "Retries": retries,
        "Delay": delay,
        "Status": status
    }
    transport_data.append(data)

    print(f"Segment {i + 1}")
    print(f"Protocol     : {protocol}")
    print(f"Port         : {port}")
    print(f"Segment Size : {segment_size} bytes")
    print(f"Retries      : {retries}")
    print(f"Delay        : {delay} ms")
    print(f"Status       : {status}")
    print("---------------------------")

print("Transport layer monitoring completed!")
```

```python
df = pd.DataFrame(transport_data)

print("Transport Layer Data")
display(df)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 7))

axes[0].plot(df["Delay"], marker="o")
axes[0].set_title("Transport Delay")
axes[0].set_ylabel("Delay (ms)")
axes[0].grid(True)
axes[1].bar(df["Segment"], df["Retries"])
axes[1].set_title("Segment Retry Count")
axes[1].set_ylabel("Retries")
axes[1].set_xlabel("Segment Number")
axes[1].grid(axis="y")

plt.tight_layout()
plt.show()
```

```python
print("=== TRANSPORT CONDITION MONITORING ===")

for reading in transport_data:
    if reading["Status"] == "Lost":
        print(f"Segment {reading['Segment']} -> DELIVERY FAILED")
    elif reading["Delay"] > 80:
        print(f"Segment {reading['Segment']} -> HIGH DELAY")
    else:
        print(f"Segment {reading['Segment']} -> NORMAL")
```

```python
import random
import pandas as pd
import matplotlib.pyplot as plt

print("=== OSI SESSION LAYER ===")
print("Virtual Session Monitoring System\\n")

session_data = []
states = ["Active", "Timed Out", "Closed"]

for i in range(10):
    duration = random.randint(10, 600)
    state = random.choices(states, weights=[7, 2, 1])[0]

    data = {
        "Session ID": f"S{i + 1:03d}",
        "User": f"User{i + 1}",
        "Duration": duration,
        "Session State": state
    }
    session_data.append(data)

    print(f"Session {i + 1}")
    print(f"Session ID : {data['Session ID']}")
    print(f"User       : {data['User']}")
    print(f"Duration   : {duration} seconds")
    print(f"State      : {state}")
    print("---------------------------")

print("Session layer monitoring completed!")
```

```python
df = pd.DataFrame(session_data)

print("Session Layer Data")
display(df)
```

```python
plt.figure(figsize=(10, 4))
plt.bar(df["Session ID"], df["Duration"])
plt.title("Session Duration Monitoring")
plt.ylabel("Duration (seconds)")
plt.xlabel("Session ID")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```python
print("=== SESSION CONDITION MONITORING ===")
for reading in session_data:
    if reading["Session State"] == "Timed Out":
        print(f"{reading['Session ID']} -> SESSION TIMEOUT")
    else:
        print(f"{reading['Session ID']} -> {reading['Session State']}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

print("=== OSI PRESENTATION LAYER ===")
print("Data Formatting and Encoding System\\n")

messages = [
    "Temperature=29.5",
    "Humidity=62",
    "Light=450",
    "Temperature=37.2",
    "Humidity=75"
]
presentation_data = []

for i, message in enumerate(messages, start=1):
    encoded = message.encode("utf-8").hex()
    decoded = bytes.fromhex(encoded).decode("utf-8")
    status = "Correct" if decoded == message else "Mismatch"

    data = {
        "Message ID": i,
        "Original Data": message,
        "Encoded Data": encoded,
        "Decoded Data": decoded,
        "Status": status
    }
    presentation_data.append(data)

    print(f"Message {i}")
    print(f"Original : {message}")
    print(f"Encoded  : {encoded}")
    print(f"Decoded  : {decoded}")
    print(f"Status   : {status}")
    print("---------------------------")

print("Presentation layer processing completed!")
```

```python
df = pd.DataFrame(presentation_data)

print("Presentation Layer Data")
display(df)
```

```python
df["Original Length"] = df["Original Data"].str.len()
df["Encoded Length"] = df["Encoded Data"].str.len()

plt.figure(figsize=(10, 4))
plt.plot(df["Message ID"], df["Original Length"], marker="o", label="Original")
plt.plot(df["Message ID"], df["Encoded Length"], marker="o", label="Hex Encoded")
plt.title("Data Length Comparison")
plt.xlabel("Message ID")
plt.ylabel("Length (characters)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print("=== PRESENTATION CONDITION MONITORING ===")

for reading in presentation_data:
    if reading["Status"] == "Correct":
        print(f"Message {reading['Message ID']} -> DATA DECODED CORRECTLY")
    else:
        print(f"Message {reading['Message ID']} -> DATA MISMATCH")
```
